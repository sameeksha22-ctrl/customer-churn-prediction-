```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# Create or load your data first
# Option 1: Load data from a file
# data = pd.read_csv('your_file.csv')  # Uncomment and modify this line to load your data

# Option 2: Create sample data for demonstration
# Here's a simple example with random data
import numpy as np
data = pd.DataFrame({
    'feature1': np.random.random(100),
    'feature2': np.random.random(100),
    'feature3': np.random.random(100),
    'target_column': np.random.choice([0, 1], size=100)  # Binary target for classification
})

# Define your feature data and target variable
X = data.drop('target_column', axis=1)  # Features
y = data['target_column']  # Target variable

# Train your random forest model
rf_model = RandomForestClassifier()
rf_model.fit(X, y)

# Get feature importance
feature_importance = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
)

# Display feature importance
print(feature_importance)
```

```python
#sort feature by importance 
feature_importance_sorted = feature_importance.sort_values(ascending = False)
feature_importance_sorted.head(10)
```

```python
# Import matplotlib.pyplot library first
import matplotlib.pyplot as plt

# visualize features importance 
plt.figure(figsize=(8,5))
feature_importance_sorted.head(10).plot(kind = 'bar')
plt.title("Top 10 features affecting customer churn  ")
plt.ylabel("Importance score")
plt.show()
```

#business insight  from feature importance 
customer with short tenure are more likely to churn 
month to month contracts increase churn risk 
high monthly charges strongly affect churn 
long term contract reduce churn significantly
